```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model
from pygame import mixer
# Load the pre-trained drowsiness detection model
model = load_model(r'D:\Drowsiness_detection_project\models\new_model.keras')
mixer.init()
sound=mixer.Sound(r'D:\Drowsiness_detection_project\alarm.wav')
# Load the pre-trained face detection DNN model
face_net = cv2.dnn.readNetFromCaffe(
    "D:/Drowsiness_detection_project/deploy.prototxt",  # Path to the prototxt file
    "D:/Drowsiness_detection_project/res10_300x300_ssd_iter_140000.caffemodel"  # Path to the caffemodel file
)

# Variables to track drowsiness score
score = 0
threshold = 10  # Trigger alarm if score exceeds threshold

# Function to preprocess input for the drowsiness detection model
def preprocess_eye(eye_frame):
    eye_frame = cv2.resize(eye_frame, (80, 80))  # Resize to the model's input size
    eye_frame = eye_frame / 255.0  # Normalize pixel values
    eye_frame = eye_frame.reshape(1, 80, 80, 3)  # Add batch dimension
    return eye_frame

# Initialize OpenCV's video capture
cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()
    if not ret:
        print("Failed to capture frame.")
        break

    # Prepare the frame for face detection
    h, w = frame.shape[:2]
    blob = cv2.dnn.blobFromImage(frame, 1.0, (300, 300), [104.0, 177.0, 123.0], False, False)

    # Perform face detection
    face_net.setInput(blob)
    detections = face_net.forward()

    for i in range(detections.shape[2]):
        confidence = detections[0, 0, i, 2]
        if confidence > 0.5:  # Confidence threshold
            box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
            (startX, startY, endX, endY) = box.astype("int")

            # Draw the bounding box for the face
            cv2.rectangle(frame, (startX, startY), (endX, endY), (255, 0, 0), 2)

            # Extract the face region
            face = frame[startY:endY, startX:endX]

            if face.size != 0:
                # Assume eyes are in the upper half of the face
                eye_region = face[0:face.shape[0]//2, :]
                
                # Resize and preprocess the eye region
                try:
                    preprocessed_eye = preprocess_eye(eye_region)
                    
                    # Predict drowsiness
                    prediction = model.predict(preprocessed_eye)
                    is_closed = prediction[0][0] > 0.6 # Adjust threshold as per your model
                    
                    if is_closed:
                        score = score = min(15, score +1)
                        cv2.putText(frame, "Closed", (startX, startY - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 255), 2)
                    else:
                        score = max(0, score - 1)  # Decrease score, ensuring it doesn't go negative
                        cv2.putText(frame, "Open", (startX, startY - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

                    # Trigger alarm if score exceeds threshold
                    if score > threshold:
                        cv2.putText(frame, "DROWSY ALERT!", (50, 50), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)
                        # Add your alarm code here (e.g., sound an alarm)
                        sound.play()
                
                except Exception as e:
                    print(f"Error processing eye frame: {e}")

    # Display the output frame
    cv2.imshow("Drowsiness Detection", frame)

    # Break on 'q' key press
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()

```
